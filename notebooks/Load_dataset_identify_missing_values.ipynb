```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# -------------------------------------------------
# 1. Load the Dataset
# -------------------------------------------------
df = pd.read_csv(
    "C:/Users/srinu/PycharmProjects/FlightFore/datasets/flights_10000_each_month_with_class_labels.csv"
)

# -------------------------------------------------
# 2. Retrieve Data in Different Ways
# -------------------------------------------------

# View the first 5 rows
print("----- First 5 Rows -----")
print(df.head())

# View the last 5 rows
print("\n----- Last 5 Rows -----")
print(df.tail())

# Display dataset shape
print("\n----- Dataset Shape -----")
print(df.shape)

# Display column names
print("\n----- Column Names -----")
print(df.columns.tolist())

# Display first 6 columns
print("\n----- First 6 Columns -----")
subset = df.iloc[:, 0:6]
print(subset)

# Display selected important columns (if available)
print("\n----- Important Flight Information -----")

important_cols = [
    "AIRLINE",
    "ORIGIN_AIRPORT",
    "DESTINATION_AIRPORT",
    "DEPARTURE_DELAY",
    "ARRIVAL_DELAY",
    "Delay"
]

available_cols = [col for col in important_cols if col in df.columns]
print(df[available_cols].head())

# -------------------------------------------------
# 3. Dataset Information
# -------------------------------------------------

print("\n----- Dataset Information -----")
print(df.info())

print("\n----- Data Types -----")
print(df.dtypes)

# -------------------------------------------------
# 4. Missing Values
# -------------------------------------------------

missing_counts = df.isnull().sum()

print("\n----- Missing Values Per Column -----")
print(missing_counts)

print("\nTotal Missing Values :", missing_counts.sum())

print("-" * 50)

# -------------------------------------------------
# 5. Duplicate Rows
# -------------------------------------------------

duplicate_rows = df[df.duplicated()]

print("----- Duplicate Rows -----")
print("Total Duplicate Rows:", len(duplicate_rows))

if len(duplicate_rows) > 0:
    print(duplicate_rows)
else:
    print("No Duplicate Rows Found.")

print("-" * 50)

# -------------------------------------------------
# 6. Statistical Summary
# -------------------------------------------------

print("\n----- Statistical Summary -----")
print(df.describe(include='all'))

# -------------------------------------------------
# 7. Missing Values Heatmap
# -------------------------------------------------

plt.figure(figsize=(12,6))

sns.heatmap(
    df.isnull(),
    cbar=False,
    yticklabels=False,
    cmap="viridis"
)

plt.title("Missing Values Heatmap")
plt.tight_layout()
plt.show()

print("\nData Retrieval and Missing Value Analysis Completed Successfully.")
```
